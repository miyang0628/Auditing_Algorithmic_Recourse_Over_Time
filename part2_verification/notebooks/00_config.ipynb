Shared paths and grayscale plotting style used across all notebooks.

```python
# 00_config.ipynb
# Shared configuration: directory paths and global grayscale plot style.
# Run this first, or %run it from other notebooks.

import os
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import seaborn as sns

# Project-relative paths (notebooks/ is the working directory)
ROOT      = os.path.abspath("..")
DATA      = os.path.join(ROOT, "data")
RAW       = os.path.join(DATA, "raw")
FIG       = os.path.join(ROOT, "results", "figures")
TAB       = os.path.join(ROOT, "results", "tables")
for d in (FIG, TAB):
    os.makedirs(d, exist_ok=True)

# Global grayscale theme for all figures
sns.set_theme(style="whitegrid", context="paper")
plt.rcParams.update({
    "font.size": 11,
    "axes.edgecolor": "0.3",
    "grid.color": "0.85",
    "axes.grid": True,
    "savefig.dpi": 600,
    "figure.dpi": 120,
})

def save_fig(fig, name):
    """Save a figure as both PNG and PDF at 600 dpi, no caption, tight bbox."""
    fig.savefig(os.path.join(FIG, name + ".png"), dpi=600, bbox_inches="tight")
    fig.savefig(os.path.join(FIG, name + ".pdf"), bbox_inches="tight")
    plt.close(fig)

print("Config loaded. DATA =", DATA)
```
